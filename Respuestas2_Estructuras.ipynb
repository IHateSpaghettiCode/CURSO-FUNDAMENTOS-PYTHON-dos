```python
#1. En un sistema de automatización industrial, un motor puede estar encendido o apagado. 
# Si la temperatura de la máquina supera los 80 grados, el motor debe apagarse automáticamente. 
# Escribir un programa que controle el estado del motor y lo apague si la temperatura supera los 80 grados.

class Motor:
    def __init__(self):
        self.estado = "apagado"  #El motor estara apagado

    def encender(self):
        self.estado = "encendido"
        print("El motor está encendido.")

    def apagar(self):
        self.estado = "apagado"
        print("El motor está apagado.")

    def verificar_temperatura(self, temperatura):
        if temperatura > 80:
            self.apagar()  #El motor se apagara si la temperatura supera los 80 grados
        else:
            print(f"La temperatura es {temperatura} grados. El motor sigue {self.estado}.")

#Apertura de if
if __name__ == "__main__":
    motor = Motor()
    motor.encender()  # Encender el motor

    while True:
        try:
            #Se hace el ingreso de la temperatura
            temperatura = float(input("Ingrese la temperatura (o escriba 'exit' para salir): "))
            motor.verificar_temperatura(temperatura)
        except ValueError:
            print("Saliendo del programa.")
            break  # Salir del bucle si el usuario ingresa un valor no numérico
```

```python
#Se calcula el descuento y precio final
def calcular_precio_matricula(valor_matricula, estrato, edad):
    if estrato == 1:
        if edad < 18:
            descuento = valor_matricula * 0.20  # 20% de descuento
        else:
            descuento = valor_matricula * 0.15  # 15% de descuento
    elif estrato == 2:
        if edad < 18:
            descuento = valor_matricula * 0.10  # 10% de descuento
        else:
            descuento = valor_matricula * 0.05  # 5% de descuento
    else:
        descuento = 0  #Sin descuento si el estrato es diferente

    precio_final = valor_matricula - descuento
    return descuento, precio_final

#Se solicita el valor de la matricula, el estrato y la edad
valor_matricula = float(input("Ingrese el valor de la matrícula: "))
estrato = int(input("Ingrese el estrato del estudiante (1 o 2): "))
edad = int(input("Ingrese la edad del estudiante: "))

#Se calcula el descuento y precio final
descuento, precio_final = calcular_precio_matricula(valor_matricula, estrato, edad)

#Se da el valor del descuento o la matricula
print(f"El valor del descuento es: {descuento:.2f}")
print(f"El precio que deberá pagar el estudiante por su matrícula es: {precio_final:.2f}")
```

```python
#Se inician las variables
total_recaudado = 0
total_descuento = 0

#Numero de personas
N = int(input("¿Cuántas personas van a ingresar? "))

#Bucle para cada persona
for i in range(N):
    #Se obtienen los datos de la persona
    estrato = int(input("Introduce el estrato (1 o 2): "))
    edad = int(input("Introduce la edad: "))
    valor_boleta = float(input("Introduce el valor de la boleta: "))
    
    #Calculo para el descuento
    if estrato == 1 and edad < 18:
        descuento = 0.20  # 20% de descuento
    elif estrato == 1 and edad >= 18:
        descuento = 0.15  # 15% de descuento
    elif estrato == 2 and edad < 18:
        descuento = 0.10  # 10% de descuento
    elif estrato == 2 and edad >= 18:
        descuento = 0.05  # 5% de descuento
    else:
        descuento = 0  # Sin descuento si no se cumple ninguna condición

    #Pago despues del descuento
    pago = valor_boleta - (valor_boleta * descuento)
    
    #Sumar al total recaudado
    total_recaudado += pago
    total_descuento += valor_boleta - pago

#Resultados (hora de la verdad)
print("Total recaudado:", total_recaudado)
print("Total descuento aplicado:", total_descuento)
```
