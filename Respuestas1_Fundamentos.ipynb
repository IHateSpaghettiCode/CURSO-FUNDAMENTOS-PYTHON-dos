```python
# 1. Calcular el índice de cosecha:
def indice_cosecha(frutos_recolectados, frutos_producidos):
    if frutos_producidos == 0:
        return 0  # Evitar división por cero
    return (frutos_recolectados / frutos_producidos) * 100

# Ejemplo de uso para el índice de cosecha
frutos_recolectados = float(input("Ingrese la cantidad de frutos recolectados: "))
frutos_producidos = float(input("Ingrese la cantidad de frutos producidos: "))
indice = indice_cosecha(frutos_recolectados, frutos_producidos)
print(f"Índice de cosecha: {indice:.2f}%")
```

```python
#Problema del promedio
#Definir para el calculo final
def calificacion_final(calificaciones_parciales, examen_final, trabajo_final):
    promedio_parciales = sum(calificaciones_parciales) / len(calificaciones_parciales)
    return (promedio_parciales * 0.55) + (examen_final * 0.30) + (trabajo_final * 0.15)

#Solicitar las calificaciones
calificaciones_parciales = []
for i in range(3):
    calificacion = float(input(f"Ingrese la calificación parcial {i + 1}: "))
    calificaciones_parciales.append(calificacion)

#Solicitar la calificacion del examen final
examen_final = float(input("Ingrese la calificación del examen final: "))

#Solicitar la calificacion del trabajo final
trabajo_final = float(input("Ingrese la calificación del trabajo final: "))

#Calculo y calificacion Final
calificacion = calificacion_final(calificaciones_parciales, examen_final, trabajo_final)
print(f"Calificación final: {calificacion:.2f}")
```

```python
#Problema de la frase y el numero de palabras
frase = input("Ingrese una frase: ")

#Aqui se cuenta el numero de palabras vaya, len cuenta los elementos de una lista y split cuenta los espacios en blanco generando la lista
numero_palabras = len(frase.split())

#Se usa el print para el resultado, la f es un formateo de cadena para que sea mas legible
print(f"Número de palabras en la frase: {numero_palabras}")
```

```python
#Problema de la frase invertida y en mayus, se pide por un input la frase al usuario
frase = input("Ingrese una frase: ")

#Se transforma la frase ingresada por medio de input, la funcion upper convierte todo en mayuscula, [::-1] es una técnica de slicing que invierte la cadena. Al usar [::-1], se crea una nueva cadena que es la original pero en orden inverso.

frase_transformada = frase.upper()[::-1]

#Mostrar el resultado
print(f"Frase en mayúsculas e invertida: {frase_transformada}")
```
